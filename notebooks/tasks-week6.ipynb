# Product Inventory

The inventory is stored in a dictionary, where each product has a stock count. The program asks for a product name and reports whether it is in stock, low in stock (fewer than 5 units), or out of stock.

```python
# The Corner Store inventory: product name -> units in stock
inventory = {
    "coffee": 12,
    "tea": 4,
    "sugar": 0,
    "milk": 8,
    "bread": 3,
    "butter": 6
}

low_stock_limit = 5

def get_stock_status(stock_count):
    """Return a label describing the stock level."""
    if stock_count == 0:
        return "OUT OF STOCK"
    elif stock_count < low_stock_limit:
        return "LOW STOCK"
    else:
        return "IN STOCK"

# Print a compact stocktake board
print("=" * 46)
print("THE CORNER STORE  |  DAILY STOCKTAKE")
print("=" * 46)
print(f"{'PRODUCT':<16}{'UNITS':>8}   {'STATUS'}")
print("-" * 46)

for name, count in inventory.items():
    print(f"{name.title():<16}{count:>8}   {get_stock_status(count)}")

print("-" * 46)
product = input("Check a product (or type its name): ").strip().lower()

if not product:
    print("Please enter a product name to check.")
elif product not in inventory:
    print(f"Sorry, '{product}' is not in the inventory.")
else:
    stock_count = inventory[product]
    status = get_stock_status(stock_count)
    print("\n" + "=" * 46)
    print(f"{product.upper()}  |  {status}")
    print(f"Units on the shelf: {stock_count}")
    print("=" * 46)
```
